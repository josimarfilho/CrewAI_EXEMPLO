### Analise de mercado para uma loja de artigo infantil o interior da Paraiba

```python
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())
```

```python
from crewai import Agent, Task, Crew
```

### Agentes
- Pesquisador de mercado
- Analista de tendencias
- Redator do relatorio

```python
pesquisador = Agent(
    role="pesquisador de mercado",
    goal="Coletar e organizar informações sobre {setor} ",
    backstory="""
    Você é um pesquisador de mercado com experiência em coletar e analisar dados de diferentes {setor}.
    Você tem um histórico comprovado de ajudar empresas a entender melhor o mercado e a tomar decisões informadas.
    Seu trabalho é garantir que as informações coletadas sejam precisas, relevantes e úteis para a empresa e estejam bem documentadas.
    Você é detalhista e tem um bom senso crítico para avaliar a qualidade das informações.
    """,
    allow_delegation=False,
    verbose=True
)
```

```python
analista = Agent(
    role="Analista de tendencias de mercado",    
    goal="Analisar e interpretar os dados coletados sobre setor {setor} e identificar tendências, oportunidades e padrões",
    backstory="""
    Você é um analista de dados com experiência em interpretar e analisar grandes volumes de dados do {setor}.
    Você tem um histórico comprovado de ajudar empresas a entender melhor os dados e a tomar decisões informadas.
    Seu trabalho é garantir que as informações analisadas sejam precisas, relevantes e úteis para a empresa.
    Você é detalhista e tem um bom senso crítico para avaliar a qualidade das informações principalmente os riscos e oportunidades.
    """,
    allow_delegation=False,
    verbose=True
)
```

```python
Redator = Agent(
    role="Redator de relatórios",
    goal="Criar um relatório claro e conciso com as informações coletadas e analisadas sobre o setor {setor}",
    backstory="""
    Você é um redator experiente com habilidade em transformar analises de mercado e criar relatórios claros e concisos.
    Você tem um histórico comprovado de ajudar empresas a comunicar informações complexas de forma simples e eficaz.
    Seu trabalho é garantir que o relatório seja bem estruturado, claro e fácil de entender.
    Você tem um bom senso crítico para avaliar a qualidade das informações e garantir que sejam apresentadas de forma clara e objetiva.
    """,
    allow_delegation=False,
    verbose=True
)
```

```python
coleta_dados = Task(
    description=(
        "1. Pesquisar e Coletar dados Atualizados sobre {setor} e organizá-los de forma estruturada."
        "2. Identificar os principais players, tendencias e estatisticas do setor."
        "3. Avaliar a qualidade e relevância das informações coletadas."
        "4. Organizar os dados de forma clara e acessível para a análise."
    ),
    expected_output="um documento estruturado com as informações coletadas sobre o setor {setor}",
    agent=pesquisador
)
```

```python
analise_tendencias = Task(
    description=(
        "1. Evaminar os dados coletados pelo pesquisador de mercado sobre o setor {setor} e identificar tendências, oportunidades e padrões."
        "2. Avaliar a qualidade e relevância das informações analisadas."
        "3. Identificar os principais riscos, tendencias e oportunidades do setor. {setor}"
        "4. Criar uma analise detalhadas as principais conclusões da análise."
    ),
    expected_output="um documento estruturado com insight e tendencias baseados nas informações analisadas sobre o setor {setor}",
    agent=analista
)
```

```python
redacao_relatorio = Task(
    description=(
        "1. Criar um relatório claro e conciso com as informações coletadas e analisadas na analise de tendencias sobre o setor {setor}."
        "2. Estruturar o relatório de forma clara e acessível para a leitura."
        "3. Garantir que o relatório seja claro, objetivo e fácil de entender."
        "4. Incluir gráficos, tabelas ou outros elementos visuais para ilustrar as informações apresentadas."
    ),
    expected_output="um relatório de analise de mercado em formato markdown, pronto para leitura e apresentação claro e conciso com as informações coletadas e analisadas sobre o setor {setor}",
    agent=Redator
)
```

```python
crew = Crew(
    agents = [pesquisador, analista, Redator],
    tasks = [coleta_dados, analise_tendencias, redacao_relatorio],
    verbose = True
)
```

```python
crew
```

```python
resultado = crew.kickoff(inputs={"setor": "Vendas de artigos infantis no sertão da paraiba"})
```
